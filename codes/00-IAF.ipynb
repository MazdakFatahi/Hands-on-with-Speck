## Single IAF Neuron
#### Constant leak LIF

```python
import sinabs.layers as sl
import torch
import matplotlib.pyplot as plt
```

```python
iaf = sl.IAF(record_states= True, spike_threshold=5. ) 
```

```python
n_steps = 500
input = (torch.rand((1, n_steps, 1))<0.05).float() # A random binary tensor with a low probability of ones
```

```python
output = iaf(input)
```

```python
iaf.recordings.keys()
```

```python
fig, axes = plt.subplots(3, 1, sharex = True, figsize = (15, 5))
axes[0].eventplot(torch.where(input)[1])
axes[1].plot(iaf.recordings['v_mem'].squeeze().numpy())
axes[2].eventplot(torch.where(output)[1])
axes[0].set_ylabel("Input events")
axes[1].set_ylabel("V_mem")
axes[2].set_ylabel("Output events")
axes[2].set_xlabel("Time-steps")
```
