# Load Model

### sinabs model

```python
import torch
sinabs_model = torch.load('NMNIST_sinabs_model.pth')
```

### DynapCNN model - DVS

```python
from sinabs.backend.dynapcnn import DynapcnnNetwork
input_shape = (2, 34, 34)
dynapcnn_model = DynapcnnNetwork(sinabs_model, input_shape=input_shape, dvs_input=True)

def config_modify_callback(devkit_cfg):
    devkit_cfg.dvs_layer.mirror.x = True
    return devkit_cfg

dynapcnn_model.to( device="speck2fdevkit:0", monitor_layers=["dvs", -1], chip_layers_ordering="auto", config_modifier=config_modify_callback)
print(f"The SNN is deployed on the core: {dynapcnn_model.chip_layers_ordering}")
```
